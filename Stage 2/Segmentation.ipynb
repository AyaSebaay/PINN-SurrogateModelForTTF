```python
import simnibs
import simnibs.segmentation
from simnibs.segmentation import charm_main
import numpy as np
import pandas as pd
import os
```

```python
def create_head_model(output_dir, t1_path, t2_path=None):
    charm_main.run(
    subject_dir = output_dir,
    T1 = t1_path,
    T2 = t2_path,
    initatlas = True,       
    segment = True,        
    create_surfaces = True, 
    mesh_image = True, 
    usesettings = "E:/Datasets/Stage 2/settings.ini"
    )
```

```python
df = pd.read_csv("E:/Datasets/Stage 2/recorrection 1.csv")
subject_id_column = 'Subject ID'
for subject_id in df[subject_id_column]:
    print(f"processing {subject_id}")
    t1 = f"E:/Datasets/Re-Creation/Co_registration/{subject_id}/t1/t1_to_mni.nii.gz"
    t2 = f"E:/Datasets/Re-Creation/Co_registration/{subject_id}/t2/t2_to_mni.nii.gz"
    m2m_dir = "E:/Datasets/Stage 2/Meshes"
    m2m_dir = os.path.join(m2m_dir, f"m2m_{subject_id}")
    os.makedirs(m2m_dir, exist_ok=True)
    create_head_model(m2m_dir , t1 , t2)
    print(f"finished {subject_id}")
```
