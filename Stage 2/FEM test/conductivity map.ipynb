```python
import nibabel as nib
import numpy as np
import random
import os
import pandas as pd
```

```python
conductivity_ranges = {
        # Original tissues
        'White-Matter': 0.12, 
        'Gray-Matter': 0.25, 
        'CSF': 1.79, 
        'Bone': 0.013, 
        'Scalp': 0.25, 
        'Eye_balls': 1, 
        'Compact_bone': 0.007, 
        'Spongy_bone': 0.03,
        'Blood': 0.65, 
        'Muscle': 0.3, 
        'Electrode': 2, 
        'gel': 0.2, 
        
        # Tumor regions
        'Necrosis': 1, 
        'Edema': 1, 
        'Enhanced_Tumor': 0.24, 
        
    }

all_labels = {
    # Original labels (from your first lookup table)
    1: conductivity_ranges['White-Matter'],
    2: conductivity_ranges['Gray-Matter'],
    3: conductivity_ranges['CSF'],
    4: conductivity_ranges['Bone'],
    5: conductivity_ranges['Scalp'],
    6: conductivity_ranges['Eye_balls'],
    7: conductivity_ranges['Compact_bone'],
    8: conductivity_ranges['Spongy_bone'],
    9: conductivity_ranges['Blood'],
    10: conductivity_ranges['Muscle'],
    2000: conductivity_ranges['Electrode'],
    2400: conductivity_ranges['gel'],
    2001: conductivity_ranges['Electrode'],
    2401: conductivity_ranges['gel'],
    
    # Tumor labels
    13: conductivity_ranges['Necrosis'],
    15: conductivity_ranges['Edema'],
    17: conductivity_ranges['Enhanced_Tumor']
}

voltage_mapping = {
    2000: -1,
    2001: 1
}
```

```python
def create_conductivity_map(seg_path, output_path):
        labeled_image = nib.load(seg_path)
        labeled_data = labeled_image.get_fdata()
        cond_data = np.zeros_like(labeled_data)
        
       
        
        # Assign conductivities
        for label, cond in all_labels.items():
            mask = labeled_data == label
            cond_data[mask] = cond
        
        
        # Save conductivity map
        cond_img = nib.Nifti1Image(cond_data ,labeled_image.affine, labeled_image.header)
        nib.save(cond_img, output_path)
        
        return cond_data
```

```python
def create_vlotage_map(base_seg_path, output_path):
    labeled_image = nib.load(base_seg_path)
    labeled_data = labeled_image.get_fdata()

    voltage_data = np.zeros_like(labeled_data)
    
    for label, voltage in voltage_mapping.items():
        mask = labeled_data == label
        voltage_data[mask] = voltage


    voltage_image = nib.Nifti1Image(voltage_data, labeled_image.affine , labeled_image.header)
    nib.save(voltage_image, output_path)

    return voltage_data
```

```python

output_dir = "D:/Results/FEM/conductavity/conductivity_map27_2.nii.gz"
voltage_output_dir = "D:/Results/FEM/conductavity/voltage_map27_2.nii.gz"



base_seg_path = "D:/Results/FEM/mesh with electrodes/test_labels27_2.nii.gz"


cond_map = create_conductivity_map(
     base_seg_path,
    output_dir,
   )

voltage_map = create_vlotage_map(base_seg_path, 
    voltage_output_dir)

print(f"Created conductivity map in {output_dir}")
```
