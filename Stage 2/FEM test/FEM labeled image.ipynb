```python
import nibabel as nib
import numpy as np
import random
import os
import pandas as pd
```

```python
tumor_mapping = {
    1: 13,  # Necrosis
    2: 15,  # Edema
    3: 17   #Enhanced Tumor
}

```

```python
def merge_segmentations(base_path, tumor_path , output_path):
        base = nib.load(base_path)
        tumor = nib.load(tumor_path)
        
       
        base_data = base.get_fdata()
        tumor_data = tumor.get_fdata()
        
        # Create merged segmentation (copy base first)
        merged = base_data.copy()
        
        
        
        # Replace tumor labels in merged data
        for old_label, new_label in tumor_mapping.items():
            merged[tumor_data == old_label] = new_label
        merged = merged.astype(np.int32)
        merged_image = nib.Nifti1Image(merged, base.affine , base.header)
        nib.save(merged_image, output_path)
        return merged
```

```python
tumor_seg_path = f"E:/Datasets/Re-Creation/Co_registration/TCGA-02-0027/seg/seg_to_mni.nii.gz"
output_dir = "D:/Results/FEM/nifti labeled/labeled_FEM-Test_TCGA-02-0027.nii.gz"

base_seg_path = "E:/Datasets/Stage 2/Meshes/m2m_TCGA-02-0027/final_tissues.nii.gz"
# Perform the merge
merged_data = merge_segmentations(base_seg_path, tumor_seg_path , output_dir)

        
print("done")
```
