```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

def load_nifti_image(file_path):
    """Load NIfTI image and return the data array."""
    img = nib.load(file_path)
    return img.get_fdata()

def apply_mask_and_calculate_averages(image1_data, image2_data, mask_data):
    
    # Define labels to keep (where we want to calculate averages)
    labels_to_keep = [13, 15, 17]
    
    # Squeeze mask if it has singleton dimensions
    mask_data = np.squeeze(mask_data)
    
    print(f"Image 1 shape: {image1_data.shape}")
    print(f"Image 2 shape: {image2_data.shape}")
    print(f"Mask shape: {mask_data.shape}")
    
    # Create a boolean mask where labels_to_keep are present
    mask_bool = np.isin(mask_data, labels_to_keep)
    
    # Create copies to avoid modifying original data
    masked_img1 = np.copy(image1_data)
    masked_img2 = np.copy(image2_data)
    
    # Set values to 0 where mask condition is False (outside our labels of interest)
    masked_img1[~mask_bool] = 0
    masked_img2[~mask_bool] = 0
    
    # Calculate average values in the masked region
    avg_img1 = np.mean(masked_img1[mask_bool])
    avg_img2 = np.mean(masked_img2[mask_bool])
    print(f"avg FEM {avg_img1}")
    print(f"avg PINN {avg_img2}")

    
    return masked_img1, masked_img2, avg_img1 ,avg_img1


image1_path = "D:/Results/FEM/Electric field energy train 5/FEM/Emagnatude_2.nii.gz"
image2_path = "D:/Results/FEM/Electric field energy train 5/PINN/Emagnatude_2.nii.gz"
mask_path = "D:/Results/FEM/nifti labeled/labeled_FEM-Test_TCGA-02-0027.nii.gz"

# Load images
print("Loading images...")
img1_data = load_nifti_image(image1_path)
img2_data = load_nifti_image(image2_path)
mask_data = load_nifti_image(mask_path)

# Apply mask and calculate averages
masked_img1, masked_img2, average1 , average2 = apply_mask_and_calculate_averages(
    img1_data, img2_data, mask_data
)

# Visualize results
#visualize_results(masked_img1, masked_img2, averages)
    
   
```
