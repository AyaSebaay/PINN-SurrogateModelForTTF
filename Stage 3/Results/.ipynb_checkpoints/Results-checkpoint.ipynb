```python
import matplotlib.pyplot as plt
import json
import os
```

```python
def load_loss_data(json_path):
    """Load previously saved loss data"""
    
    with open(json_path, 'r') as f:
        loss_data = json.load(f)
    print(f"Loss data loaded from {json_path}")
    return loss_data
    
```

```python
def plot_loss_curves(loss_data):
    """Plot training and validation loss curves"""
    epochs = loss_data['epochs']
    train_losses = loss_data['train_losses']
    val_losses = loss_data['val_losses']
    train_physics = loss_data['train_physics_losses']
    train_boundary = loss_data['train_boundary_losses']
    val_physics = loss_data['val_physics_losses']
    val_boundary = loss_data['val_boundary_losses']
    
    # Create subplots
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
    
    # Plot 1: Total Loss
    ax1.plot(epochs, train_losses, label='Training Loss', linewidth=2)
    ax1.plot(epochs, val_losses, label='Validation Loss', linewidth=2)
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.set_title('Total Loss Curve')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    ax1.set_yscale('log')  # Log scale for better visualization
    
    # Plot 2: Physics Loss
    ax2.plot(epochs, train_physics, label='Training Physics Loss', linewidth=2)
    ax2.plot(epochs, val_physics, label='Validation Physics Loss', linewidth=2)
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Physics Loss')
    ax2.set_title('Physics Loss Curve')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    ax2.set_yscale('log')
    
    # Plot 3: Boundary Loss
    ax3.plot(epochs, train_boundary, label='Training Boundary Loss', linewidth=2)
    ax3.plot(epochs, val_boundary, label='Validation Boundary Loss', linewidth=2)
    ax3.set_xlabel('Epoch')
    ax3.set_ylabel('Boundary Loss')
    ax3.set_title('Boundary Loss Curve')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    ax3.set_yscale('log')
    
    # Plot 4: Loss Components (last 100 epochs for detail)
    if len(epochs) > 100:
        last_epochs = epochs[-100:]
        last_train_total = train_losses[-100:]
        last_val_total = val_losses[-100:]
    else:
        last_epochs = epochs
        last_train_total = train_losses
        last_val_total = val_losses
    
    ax4.plot(last_epochs, last_train_total, label='Training Total', linewidth=2)
    ax4.plot(last_epochs, last_val_total, label='Validation Total', linewidth=2)
    ax4.set_xlabel('Epoch')
    ax4.set_ylabel('Loss')
    ax4.set_title('Recent Loss Trend (Last 100 Epochs)')
    ax4.legend()
    ax4.grid(True, alpha=0.3)
    ax4.set_yscale('log')
    
    plt.tight_layout()
    
    if save_dir:
        plt.savefig(os.path.join(save_dir, 'loss_curves.png'), dpi=300, bbox_inches='tight')
        plt.savefig(os.path.join(save_dir, 'loss_curves.pdf'), bbox_inches='tight')
    
    plt.show()
    
    return fig
```

```python
results_dir = ""
loss_data = load_loss_data(results_dir)
plot_loss_curves(loss_data)
```
