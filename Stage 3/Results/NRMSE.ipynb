```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

def load_nifti_image(file_path):
    """Load NIfTI image and return the data array."""
    img = nib.load(file_path)
    return img.get_fdata()

Ex_FEM_Path = "D:/Results/FEM/Electric field energy train 5/FEM/Ex_2.nii.gz"
Ey_FEM_Path = "D:/Results/FEM/Electric field energy train 5/FEM/Ey_2.nii.gz"
Ez_FEM_Path = "D:/Results/FEM/Electric field energy train 5/FEM/Ez_2.nii.gz"
Ex_PINN_Path = "D:/Results/FEM/Electric field energy train 5/PINN/Ex_2.nii.gz"
Ey_PINN_Path = "D:/Results/FEM/Electric field energy train 5/PINN/Ey_2.nii.gz"
Ez_PINN_Path = "D:/Results/FEM/Electric field energy train 5/PINN/Ez_2.nii.gz"

E_magnitude_FEM_Path = "D:/Results/FEM/Electric field energy train 5/FEM/Emagnatude_2.nii.gz"
E_magnitude_PINN_Path = "D:/Results/FEM/Electric field energy train 5/PINN/Emagnatude_2.nii.gz"

Ex_FEM = load_nifti_image(Ex_FEM_Path)
Ey_FEM = load_nifti_image(Ey_FEM_Path)
Ez_FEM = load_nifti_image(Ez_FEM_Path)
Ex_PINN = load_nifti_image(Ex_PINN_Path)
Ey_PINN = load_nifti_image(Ey_PINN_Path)
Ez_PINN = load_nifti_image(Ez_PINN_Path)
E_magnitude_FEM = load_nifti_image(E_magnitude_FEM_Path)
E_magnitude_PINN = load_nifti_image(E_magnitude_PINN_Path)

Ex_diff = Ex_PINN - Ex_FEM
Ey_diff = Ey_PINN - Ey_FEM
Ez_diff = Ez_PINN - Ez_FEM

diff_mag = np.sqrt((Ex_diff)**2 + (Ey_diff)**2 + (Ez_diff)**2)
fem_mag = np.sqrt(Ex_FEM**2 + Ey_FEM**2 + Ez_FEM**2)
pinn_mag = np.sqrt(Ex_PINN**2 + Ey_PINN**2 + Ez_PINN**2)

nonzero_mask = (fem_mag > 0)
nmse = np.mean(diff_mag[nonzero_mask])/ max(E_magnitude_FEM[nonzero_mask])

print(f"Mean Relative Error: {nmse}") 

```
