```python
import nibabel as nib
import matplotlib.pyplot as plt

# Load the T1 image
t1_img = nib.load('E:/Datasets/co-registered data/TCGA-08-0359/t1/t1_to_mni.nii.gz')
t2_img = nib.load('E:/Datasets/co-registered data/TCGA-08-0359/t2/t2_to_mni.nii.gz')
t1_data = t1_img.get_fdata()
t2_data = t2_img.get_fdata()
# Load the segmentation mask
seg_img = nib.load('E:/Datasets/co-registered data/TCGA-08-0359/seg/seg_to_mni.nii.gz')
seg_data = seg_img.get_fdata()



# Choose a slice to display (e.g., middle slice along the z-axis)
slice_idx = t1_data.shape[2] // 2
t1_slice = t1_data[:, :, slice_idx]

slice2_idx = t2_data.shape[2] // 2
t2_slice = t2_data[:, :, slice_idx]

plt.imshow(t1_slice, cmap='gray')
plt.title('T1 Image Slice')
plt.axis('off')
plt.show()

# Choose the same slice for the segmentation mask
seg_slice = seg_data[:, :, slice_idx]

# Display the T1 slice
plt.imshow(t1_slice, cmap='gray')

# Overlay the segmentation mask with transparency
plt.imshow(seg_slice, cmap='jet', alpha=0.5)  # 'jet' colormap for the mask
plt.title('T1 Image with Segmentation Overlay')
plt.axis('off')
plt.show()

plt.imshow(t2_slice, cmap='gray')
plt.title('T2 Image Slice')
plt.axis('off')
plt.show()

print(seg_slice)
```
