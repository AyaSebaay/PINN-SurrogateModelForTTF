```python
import pandas as pd

# Load the first CSV file
df1 = pd.read_csv('E:/Datasets/updated_file.csv')

# Load the second CSV file
df2 = pd.read_csv('E:/Datasets/manifest-DICOM-GLIOMA-SEG_2024/metadata - Copy.csv')

# Merge the DataFrames on the specified columns
merged_df = pd.merge(df1, df2, left_on='File Path', right_on='File Location', how='left')

# Extract the "Series UID" from the merged DataFrame and add it to the first DataFrame
df1['Series UID'] = merged_df['Series UID']

# Save the resulting DataFrame to a new CSV file
df1.to_csv('E:/Datasets/Series UID/correctPathWithSeriesUID.csv', index=False)
```
