```python
# i have created this code because i rerun the data selection with different criteria of selectiong T2
# so in order to not repeat the coregistration for same selected scan i created this 
#to only select the different T2 scan data

import pandas as pd
import os
import shutil

# File paths
file1_path = "E:/Datasets/Selected Data 1/selected_data.csv"  
file2_path = "E:/Datasets/Re-Creation/Selected Data/selected_data with SEG.csv"  
output_csv_path = 'E:/Datasets/Re-Creation/differences.csv'  

# Read both CSV files
df1 = pd.read_csv(file1_path)
df2 = pd.read_csv(file2_path)

# Ensure the dataframes have the same length
if len(df1) != len(df2):
    print("Error: The CSV files have different row counts.")
    exit()

# Find rows where 'T2 Series Description' differs
different_rows_mask = df1['T2 Series Description'] != df2['T2 Series Description']
different_rows = df2[different_rows_mask]

# Save the different rows to a new CSV file
different_rows.to_csv(output_csv_path, index=False)
print(f"Saved {len(different_rows)} differing rows to {output_csv_path}")

# Process similar rows - copy folders
similar_rows_mask = df1['T2 Series Description'] == df2['T2 Series Description']
similar_rows = df1[similar_rows_mask]


subject_id_column = 'Subject ID'
for subject_id in similar_rows[subject_id_column]:
    source_folder = f"E:/Datasets/Coregisterd-rigid/{subject_id}"
    dest_path = f"E:/Datasets/Re-Creation/Co_registration/{subject_id}"
    shutil.copytree(source_folder, dest_path)
    print(f"Copied {source_folder} to {dest_path}")
    
```
