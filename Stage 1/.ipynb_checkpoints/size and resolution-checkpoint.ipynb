```python
import nibabel as nib

# Load the NIFTI file
img = nib.load('E:/Datasets/mni.nii.gz')  # or 'your_image.nii.gz' for compressed files

# Get the image data as a numpy array
data = img.get_fdata()

# Get the image dimensions (size)
dimensions = data.shape
print(f"Image dimensions (size): {dimensions}")

# Get the voxel resolution (mm)
header = img.header
resolution = header.get_zooms()
print(f"Voxel resolution (mm): {resolution}")

# If you want them separately:
print(f"\nDetailed information:")
print(f"X dimension: {dimensions[0]} voxels, resolution: {resolution[0]} mm")
print(f"Y dimension: {dimensions[1]} voxels, resolution: {resolution[1]} mm")
print(f"Z dimension: {dimensions[2]} voxels, resolution: {resolution[2]} mm")

# For 4D images (like fMRI), you'll have a 4th dimension:
if len(dimensions) > 3:
    print(f"Time points: {dimensions[3]}, TR: {resolution[3]} sec")
```
