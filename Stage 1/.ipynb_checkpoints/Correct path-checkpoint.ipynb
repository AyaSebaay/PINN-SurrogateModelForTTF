```python
import pandas as pd

# Load the first CSV file (incorrect paths)
first_csv = "E:/Datasets/TCGA_DSO_Key (2).csv"
first_df = pd.read_csv(first_csv)

# Load the second CSV file (correct paths)
second_csv = "E:/Datasets/manifest-DICOM-GLIOMA-SEG_2024/metadata.csv"
second_df = pd.read_csv(second_csv)

# Function to clean Series Description
def clean_series_description(description):
    # Remove "glioma segmentation" or "glioma segmentation corrected"
    description = description.replace("Glioma Segmentation Corrected", "").replace("Glioma Segmentation", "")
    # Remove leading/trailing spaces and replace spaces with ""
    description = description.strip().replace(" ", "__")
    return description

# Iterate through each row in the first file
for index, row in second_df.iterrows():
    subject_id = row["Subject ID"]
    series_description = row["Series Description"]
    print(f"Processing Subject ID: {subject_id}, Series Description: {series_description}...")

    # Check if the description contains "corrected"
    contains_corrected = "corrected" in series_description.lower()

    # Clean the Series Description
    cleaned_description = clean_series_description(series_description)
    print(f"cleaned description: {cleaned_description}")

    # Find matching rows in the second file
    matching_rows = first_df[(first_df["patient_id"] == subject_id) & 
                             (first_df["filepath"].str.contains(cleaned_description, case=False, na=False))]
    
    # Filter based on the presence of "corrected"
    if contains_corrected:
        matching_rows = matching_rows[matching_rows["filepath"].str.contains("corrected", case=False, na=False)]
    else:
        matching_rows = matching_rows[~matching_rows["filepath"].str.contains("corrected", case=False, na=False)]
    
    if not matching_rows.empty:
        correct_path = row["File Location"]
        first_df.loc[matching_rows.index, "File Path"] = correct_path
        
        print(f"Updated file path to: {correct_path}")
    else:
        print(f"No matching file path found for Subject ID: {subject_id}, Series Description: {cleaned_description}")

# Save the updated first file
#updated_first_csv = "E:/Datasets/corrected path.csv"
#first_df.to_csv(updated_first_csv, index=False)

print("File path correction complete!")
print(f"Updated first file saved to: {updated_first_csv}")
```
