```python
import os
import pandas as pd
import subprocess

# Load metadata file
original_csv = "E:/Datasets/manifest-21CCx8fx964169794518489879/metadata.csv"
metadata = pd.read_csv(original_csv)

# Base directories
dicom_base_dir = "E:/Datasets/manifest-21CCx8fx964169794518489879"
nifti_base_dir = "E:/Datasets/Re-Creation/Selected Data"
os.makedirs(nifti_base_dir, exist_ok=True)

# Initialize a list to store selected data
selected_data = []

# Function to convert DICOM to NIFTI using dcm2niix
def convert_dicom_to_nifti(dicom_dir, nifti_dir, output_name):
    os.makedirs(nifti_dir, exist_ok=True)
   
    # Run dcm2niix
    command = f'dcm2niix -z y -o "{nifti_dir}" -f {output_name} "{dicom_dir}"'
    print(f"Running command: {command}")  # Debugging: Print the command
    try:
        subprocess.run(command, check=True)
    except subprocess.CalledProcessError as e:
        print(f"Error: dcm2niix failed with exit code {e.returncode}")
        return None
    # Find the generated NIFTI file
    for file in os.listdir(nifti_dir):
        if file.startswith(output_name) and file.endswith(".nii.gz"):
            return os.path.join(nifti_dir, file)
    return None



# Iterate through each subject
for subject_id, subject_data in metadata.groupby("Subject ID"):
    print(f"Processing {subject_id}...")

    # Create subject directories
    
    subject_nifti_dir = os.path.join(nifti_base_dir, f"{subject_id}")
    os.makedirs(subject_nifti_dir, exist_ok=True)

    # Select T1 (or FSPGR)
    t1_data = subject_data[subject_data["Series Description"].str.contains(
        "T1|SPGR|VOLUMETRIC AXIAL Gd|MP RAGE AXIAL VOLUMETRIC", case=False, na=False)]
    if not t1_data.empty:
        t1_series = t1_data.iloc[0]  # Select the first T1/FSPGR series
        t1_dicom_dir = os.path.join(dicom_base_dir ,t1_series["File Location"])
        t1_dicom_dir = t1_dicom_dir.replace(".\\","/")
        t1_dicom_dir = t1_dicom_dir.replace("\\","/")
        t1_dicom_dir = t1_dicom_dir.replace("//","/")
        t1_nifti_dir = os.path.join(subject_nifti_dir, "T1")
        t1_nifti_path = convert_dicom_to_nifti(t1_dicom_dir, t1_nifti_dir, "t1")
    else:
        print(f"No T1/FSPGR found for {subject_id}. Skipping...")
        continue

    # Select T2 (or FLAIR)
    t2_data = subject_data[subject_data["Series Description"].str.contains(
        "T2", case=False, na=False) &
    ~subject_data["Series Description"].str.contains("FLAIR", case=False, na=False)]

    if t2_data.empty: #use FLAIR only if no other T2 available (make sure it is not T1 FLAIR)
        t2_data = subject_data[subject_data["Series Description"].str.contains(
        "FLAIR", case=False, na=False) &
        ~subject_data["Series Description"].str.contains("T1", case=False, na=False)]
        
        
    if not t2_data.empty:
        t2_series = t2_data.iloc[0]  # Select the first T2/FLAIR series
        t2_dicom_dir = os.path.join(dicom_base_dir ,t2_series["File Location"])
        t2_dicom_dir = t2_dicom_dir.replace(".\\","/")
        t2_dicom_dir = t2_dicom_dir.replace("\\","/")
        t2_dicom_dir = t2_dicom_dir.replace("//","/")
        t2_nifti_dir = os.path.join(subject_nifti_dir, "T2")
        t2_nifti_path = convert_dicom_to_nifti(t2_dicom_dir, t2_nifti_dir, "t2")
    else:
        print(f"No T2 found for {subject_id}. Skipping T2...")
        t2_nifti_path = None

    # Add selected data to the list
    selected_data.append({
        "Subject ID": subject_id,
        "T1 Series ID": t1_series["Series UID"] if not t1_data.empty else None,
        "T1 Series Description": t1_series["Series Description"] if not t1_data.empty else None,
        "T1 File Path": t1_nifti_path if not t1_data.empty else None,
        "T2 Series ID": t2_series["Series UID"] if not t2_data.empty else None,
        "T2 Series Description": t2_series["Series Description"] if not t2_data.empty else None,
        "T2 File Path": t2_nifti_path if not t2_data.empty else None,
    })

# Save the selected data to a new csv file
selected_df = pd.DataFrame(selected_data)
selected_df.to_csv(os.path.join(nifti_base_dir, "selected_data.csv"), index=False)

print("Dataset organization and conversion complete!")
```
